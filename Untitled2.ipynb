```python
# Analisador de Imagens Médicas — Entrega 2
# Google Colab / Jupyter Notebook
# Pré-processamento com Python e OpenCV

# CÉLULA 1 — Instalação das bibliotecas
!pip -q install opencv-python-headless ipywidgets

# CÉLULA 2 — Importações
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import display, clear_output
import ipywidgets as widgets

print("Bibliotecas carregadas com sucesso.")

# CÉLULA 3 — Carregar a imagem
# Execute esta célula e selecione uma imagem PNG, JPG ou JPEG do computador.
uploaded = files.upload()

if not uploaded:
    raise ValueError("Nenhum arquivo foi enviado.")

nome_arquivo = next(iter(uploaded))
bytes_imagem = np.frombuffer(uploaded[nome_arquivo], np.uint8)
imagem_bgr = cv2.imdecode(bytes_imagem, cv2.IMREAD_COLOR)

if imagem_bgr is None:
    raise ValueError("Não foi possível abrir a imagem. Tente PNG ou JPG.")

imagem_rgb = cv2.cvtColor(imagem_bgr, cv2.COLOR_BGR2RGB)
imagem_cinza = cv2.cvtColor(imagem_bgr, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(7, 5))
plt.imshow(imagem_rgb)
plt.title("Imagem original")
plt.axis("off")
plt.show()

# CÉLULA 4 — Função de pré-processamento
def processar_imagem(tipo_filtro="Gaussiano",
                     tamanho_kernel=5,
                     aplicar_clahe=True,
                     intensidade_clahe=2.0,
                     divisao_clahe=8):
    """
    Aplica filtro opcional para reduzir ruído e CLAHE para realçar
    o contraste local. Retorna a imagem processada em escala de cinza.
    """
    resultado = imagem_cinza.copy()

    if tipo_filtro == "Gaussiano":
        resultado = cv2.GaussianBlur(
            resultado, (tamanho_kernel, tamanho_kernel), 0
        )
    elif tipo_filtro == "Mediana":
        resultado = cv2.medianBlur(resultado, tamanho_kernel)
    elif tipo_filtro == "Nenhum":
        pass

    if aplicar_clahe:
        clahe = cv2.createCLAHE(
            clipLimit=float(intensidade_clahe),
            tileGridSize=(int(divisao_clahe), int(divisao_clahe))
        )
        resultado = clahe.apply(resultado)

    return resultado

# CÉLULA 5 — Mostrar as etapas do processamento
def mostrar_comparacao(tipo_filtro="Gaussiano",
                       tamanho_kernel=5,
                       aplicar_clahe=True,
                       intensidade_clahe=2.0,
                       divisao_clahe=8):
    resultado = processar_imagem(
        tipo_filtro=tipo_filtro,
        tamanho_kernel=tamanho_kernel,
        aplicar_clahe=aplicar_clahe,
        intensidade_clahe=intensidade_clahe,
        divisao_clahe=divisao_clahe
    )

    fig, eixos = plt.subplots(1, 3, figsize=(16, 5))
    eixos[0].imshow(imagem_rgb)
    eixos[0].set_title("1. Imagem original")
    eixos[1].imshow(imagem_cinza, cmap="gray")
    eixos[1].set_title("2. Escala de cinza")
    eixos[2].imshow(resultado, cmap="gray")
    eixos[2].set_title("3. Após pré-processamento")

    for ax in eixos:
        ax.axis("off")

    plt.tight_layout()
    plt.show()

    print("Parâmetros usados:")
    print(f"- Filtro: {tipo_filtro}")
    if tipo_filtro != "Nenhum":
        print(f"- Kernel: {tamanho_kernel} x {tamanho_kernel}")
    print(f"- CLAHE: {'ativado' if aplicar_clahe else 'desativado'}")
    if aplicar_clahe:
        print(f"- clipLimit: {intensidade_clahe}")
        print(f"- Divisões locais: {divisao_clahe} x {divisao_clahe}")

    return resultado

resultado_final = mostrar_comparacao()

# CÉLULA 6 — Controles interativos
# Altere as opções e observe como a imagem muda.
filtro_widget = widgets.Dropdown(
    options=["Nenhum", "Gaussiano", "Mediana"],
    value="Gaussiano",
    description="Filtro:"
)
kernel_widget = widgets.SelectionSlider(
    options=[3, 5, 7, 9],
    value=5,
    description="Kernel:"
)
clahe_widget = widgets.Checkbox(
    value=True,
    description="Aplicar CLAHE"
)
intensidade_widget = widgets.FloatSlider(
    value=2.0, min=1.0, max=4.0, step=0.5,
    description="Intensidade:"
)
divisao_widget = widgets.SelectionSlider(
    options=[4, 6, 8, 10, 12],
    value=8,
    description="Divisões:"
)

painel = widgets.VBox([
    filtro_widget,
    kernel_widget,
    clahe_widget,
    intensidade_widget,
    divisao_widget
])

saida = widgets.interactive_output(
    mostrar_comparacao,
    {
        "tipo_filtro": filtro_widget,
        "tamanho_kernel": kernel_widget,
        "aplicar_clahe": clahe_widget,
        "intensidade_clahe": intensidade_widget,
        "divisao_clahe": divisao_widget
    }
)

display(painel, saida)

# CÉLULA 7 — Salvar a imagem processada
# Esta célula usa os parâmetros atuais dos controles interativos.
resultado_para_salvar = processar_imagem(
    tipo_filtro=filtro_widget.value,
    tamanho_kernel=kernel_widget.value,
    aplicar_clahe=clahe_widget.value,
    intensidade_clahe=intensidade_widget.value,
    divisao_clahe=divisao_widget.value
)

caminho_saida = "/content/imagem_processada.png"
cv2.imwrite(caminho_saida, resultado_para_salvar)
print("Imagem salva em:", caminho_saida)

files.download(caminho_saida)

# CÉLULA 8 — Conclusão para o relatório/apresentação
print("""
Nesta primeira versão, a aplicação permite carregar uma imagem,
visualizar sua conversão para escala de cinza, reduzir ruídos com
filtros Gaussiano ou de mediana e realçar o contraste local com CLAHE.
Os controles interativos permitem comparar parâmetros diferentes.
A segmentação será desenvolvida em uma etapa posterior.
Este protótipo é acadêmico e não realiza diagnóstico médico.
""")
```
